```python
import os
from pathlib import Path

if Path.cwd().name == "notebooks":
    os.chdir("..")

from dementia_speech.io.textgrid import load_word_tier
from dementia_speech.tasks.phonemic_fluency.scoring import score_trial, compute_metrics
```

```python
textgrid_path = Path("data/raw/textgrids/PFT/PROCESS-2_rec__039__PFT.TextGrid")
df = load_word_tier(textgrid_path)
print(f"Loaded {len(df)} words")
df.head(40)
```

```python
scored_df = score_trial(df, target_letter="p")
scored_df.head(50)
```

```python
scored_df['classification'].value_counts()
```

```python
metrics = compute_metrics(scored_df)
metrics
```
